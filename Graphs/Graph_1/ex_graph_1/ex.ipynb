```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

define state schema | define first node (node has state as input and output)

```python
class AgentState(TypedDict):
    message: str

def compliment(state: AgentState) -> AgentState:
    """A simple node that adds a compliment message to the state."""
    state["message"] = "You are awesome, " + state["message"] + "!"
    return state
```

define the graph structure | add start n end node | connect to first node | compile graph for errors

```python
graph = StateGraph(AgentState)
graph.add_node("compliment_node", compliment)
graph.set_entry_point("compliment_node")
graph.set_finish_point("compliment_node")
graph_app = graph.compile()
```

invoke the graph

```python
result = graph_app.invoke({"message": "Alice"})  # invoke the graph with the initial state
```

print results
